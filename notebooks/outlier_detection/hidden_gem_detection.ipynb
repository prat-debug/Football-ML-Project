```python
# ============================================================
# HIDDEN GEM DETECTION
# PART 1 - IMPORT LIBRARIES
# ============================================================

import warnings
warnings.filterwarnings("ignore")

from pathlib import Path

import numpy as np
import pandas as pd

from sklearn.preprocessing import (
    LabelEncoder,
    StandardScaler
)
```

```python
# ============================================================
# PROJECT PATHS
# ============================================================

PROJECT_ROOT = Path.cwd().resolve().parent.parent

DATA_DIR = PROJECT_ROOT / "data"

RAW_DIR = DATA_DIR / "cleaned"

RESULTS_DIR = PROJECT_ROOT / "results" / "hidden_gem"

MODELS_DIR = PROJECT_ROOT / "models"

RESULTS_DIR.mkdir(parents=True, exist_ok=True)
MODELS_DIR.mkdir(parents=True, exist_ok=True)

DATASET = RAW_DIR / "clean_player_features.csv"

print("Dataset :", DATASET)
print("Results :", RESULTS_DIR)
print("Models  :", MODELS_DIR)
```

```python
# ============================================================
# LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET)

print("Dataset Loaded Successfully")

print("Shape :", df.shape)

df.head()
```

```python
# ============================================================
# FEATURE SELECTION
# ============================================================

FEATURES = [

    "matches_played",

    "total_minutes",

    "total_goals",

    "total_assists",

    "goals_per_match",

    "assists_per_match",

    "minutes_per_match",

    "goals_per_90",

    "assists_per_90",

    "goal_contributions_per_90",

    "estimated_xg",

    "estimated_xa",

    "age",

    "height_in_cm",

    "preferred_foot",

    "position_group",

    "goals_score",

    "assists_score",

    "experience_score",

    "xg_score",

    "xa_score",

    "performance_score",

    "physicality_score",

    "talent_score"

]

X = df[FEATURES].copy()

print("Selected Features")

print(X.columns.tolist())
```

```python
# ============================================================
# LABEL ENCODING
# ============================================================

encoders = {}

categorical_columns = [

    "preferred_foot",

    "position_group"

]

for column in categorical_columns:

    encoder = LabelEncoder()

    X[column] = encoder.fit_transform(
        X[column]
    )

    encoders[column] = encoder

print("Categorical Features Encoded")
```

```python
# ============================================================
# FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

print("Scaling Completed")

X_scaled.head()
```

```python
# ============================================================
# SAVE PREPROCESSED FEATURES
# ============================================================

preprocessed_file = (
    RESULTS_DIR /
    "hidden_gem_preprocessed.csv"
)

X_scaled.to_csv(
    preprocessed_file,
    index=False
)

print("Saved Successfully")

print(preprocessed_file)
```

```python
# ============================================================
# PART 2 - DIMENSIONALITY REDUCTION
# ============================================================

from sklearn.decomposition import PCA
from sklearn.decomposition import KernelPCA
```

```python
# ============================================================
# FIND OPTIMAL PCA COMPONENTS
# ============================================================

pca_variance = PCA(n_components=0.95)

X_linear = pca_variance.fit_transform(X_scaled)

print("=" * 60)

print("LINEAR PCA")

print("=" * 60)

print("Original Features :", X_scaled.shape[1])

print("Selected Components :", pca_variance.n_components_)

print("Explained Variance :")

print(round(
    np.sum(
        pca_variance.explained_variance_ratio_
    ),
    4
))
```

```python
# ============================================================
# CREATE LINEAR PCA DATAFRAME
# ============================================================

linear_columns = [

    f"PC_{i+1}"

    for i in range(

        X_linear.shape[1]

    )

]

X_linear_df = pd.DataFrame(

    X_linear,

    columns=linear_columns

)

X_linear_df.head()
```

```python
# ============================================================
# KERNEL PCA
# ============================================================

kernel_pca = KernelPCA(

    n_components=X_linear.shape[1],

    kernel="rbf",

    gamma=None,

    random_state=42

)

X_kernel = kernel_pca.fit_transform(

    X_scaled

)

print("=" * 60)

print("KERNEL PCA")

print("=" * 60)

print("Components :", X_kernel.shape[1])
```

```python
# ============================================================
# CREATE KERNEL PCA DATAFRAME
# ============================================================

kernel_columns = [

    f"KPC_{i+1}"

    for i in range(

        X_kernel.shape[1]

    )

]

X_kernel_df = pd.DataFrame(

    X_kernel,

    columns=kernel_columns

)

X_kernel_df.head()
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.cluster import KMeans

from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

import pandas as pd
import numpy as np
```

```python
def evaluate_model(X, labels):

    unique_labels = np.unique(labels)

    if len(unique_labels) < 2:
        return None

    try:

        silhouette = silhouette_score(X, labels)

        db_index = davies_bouldin_score(X, labels)

        ch_index = calinski_harabasz_score(X, labels)

    except:

        silhouette = np.nan
        db_index = np.nan
        ch_index = np.nan

    outlier_percentage = (
        np.sum(labels == -1)
        / len(labels)
    ) * 100

    return {

        "Silhouette Score": silhouette,

        "Davies-Bouldin Index": db_index,

        "Calinski-Harabasz Index": ch_index,

        "Outlier Percentage": outlier_percentage

    }
```

```python
comparison_results = []
```

```python
datasets = {

    "Linear PCA": X_linear,

    "Kernel PCA": X_kernel

}
```

```python
isolation_grid = {

    "n_estimators": [100, 200],

    "contamination": [0.02, 0.05, 0.10],

    "max_samples": ["auto", 0.8],

    "random_state": [42]

}
```

```python
for dataset_name, X in datasets.items():

    print(f"\nRunning Isolation Forest on {dataset_name}")

    for n_estimator in isolation_grid["n_estimators"]:

        for contamination in isolation_grid["contamination"]:

            for sample in isolation_grid["max_samples"]:

                model = IsolationForest(

                    n_estimators=n_estimator,

                    contamination=contamination,

                    max_samples=sample,

                    random_state=42

                )

                labels = model.fit_predict(X)

                metrics = evaluate_model(X, labels)

                if metrics is None:
                    continue

                comparison_results.append({

                    "Model": "Isolation Forest",

                    "Dataset": dataset_name,

                    "Parameters": {

                        "n_estimators": n_estimator,

                        "contamination": contamination,

                        "max_samples": sample

                    },

                    **metrics

                })
```

```python
lof_grid = {

    "n_neighbors": [10,20,30,40],

    "contamination": [0.02,0.05,0.10]
}
```

```python
for dataset_name, X in datasets.items():

    print(f"\nRunning LOF on {dataset_name}")

    for neighbors in lof_grid["n_neighbors"]:

        for contamination in lof_grid["contamination"]:

            model = LocalOutlierFactor(

                n_neighbors=neighbors,

                contamination=contamination

            )

            labels = model.fit_predict(X)

            metrics = evaluate_model(X, labels)

            if metrics is None:
                continue

            comparison_results.append({

                "Model":"LOF",

                "Dataset":dataset_name,

                "Parameters":{

                    "n_neighbors":neighbors,

                    "contamination":contamination

                },

                **metrics

            })
```

```python
kmeans_grid = {

    "clusters":[2,3,4,5,6,7,8],

    "init":["k-means++"],

    "n_init":[10],

    "random_state":[42]
}
```

```python
for dataset_name, X in datasets.items():

    print(f"\nRunning KMeans on {dataset_name}")

    for cluster in kmeans_grid["clusters"]:

        model = KMeans(

            n_clusters=cluster,

            init="k-means++",

            n_init=10,

            random_state=42

        )

        labels = model.fit_predict(X)

        metrics = evaluate_model(X, labels)

        if metrics is None:
            continue

        comparison_results.append({

            "Model":"KMeans",

            "Dataset":dataset_name,

            "Parameters":{

                "clusters":cluster

            },

            **metrics

        })
```

```python
comparison_df = pd.DataFrame(comparison_results)

comparison_df.head()
```

```python
comparison_df = comparison_df.sort_values(

    by=[

        "Silhouette Score",

        "Calinski-Harabasz Index",

        "Davies-Bouldin Index"

    ],

    ascending=[False,False,True]

)

comparison_df
```
